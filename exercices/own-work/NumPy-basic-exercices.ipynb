```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# test none are zeros
def none_are_zeros(arr):
    """takes an array and returns True if no elements are zeros"""
    return np.all(arr != 0) # np.all() vérifie si tous les éléments satisfont une condition logique (ici d'être différents de zéro)

x = np.array([1, 2, 3, 4])
y = np.array([0, 1, 2, 3])
z = np.array(['a', ''])
print(none_are_zeros(x),'\n',none_are_zeros(y),'\n',none_are_zeros(z))
```

```python
# test all are zeros
def all_are_zeros(arr):
    """takes an array and returns True if no elements are different than 0"""
    return np.all(arr == 0) # np.all() vérifie si tous les éléments satisfont une condition logique (ici d'être différents de zéro)

x = np.array([1, 2, 3, 4])
y = np.array([0, 1, 2, 3])
z = np.array([0, 0, 0, 0])
print(none_are_zeros(x),'\n',none_are_zeros(y),'\n',none_are_zeros(z))
```

```python
def finitness(arr):
    """test a given array element-wise for finiteness (not infinity or not a number)"""
    return np.all(np.isfinite(arr)) #The function returns a boolean array of the same shape as 'a', with True for finite values (1 and 0) and False for non-finite values (NaN and Inf)

x = np.array([1, 2, 3, 4])
y = np.array([np.nan, 1, 2, 3])
z = np.array([1, 3, np.inf])

print(finitness(x), finitness(y), finitness(z))
```
