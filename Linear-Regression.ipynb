```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('/home/lilouan/ml-workspace/MLcourse/data/global-temperatures.csv', comment='#', header=None, sep=' ')
data.columns = ['year', 'temp']
# plt.figure(figsize=(10, 5))
plt.scatter(data['year'], data['temp'])
plt.xlabel('Year', fontsize=12)
plt.ylabel('Temperature Anomaly (°C)', fontsize=12)
plt.title('Global Annual Temperatures (1880 - 2017)', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
```

```python
# calculating the fit
x = data['year'].to_numpy()
y = data['temp'].to_numpy()

#creating the design matrix X
X = np.c_[np.ones(len(x)), x]

#calculatint predicted beta
b = np.linalg.inv(X.T @ X ) @ X.T @ y 
print(b)
```
